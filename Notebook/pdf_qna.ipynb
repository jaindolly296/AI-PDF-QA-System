```python
# Create and save this code into app.py file
%%writefile app.py

# Import Streamlit library for UI
import streamlit as st

# Import Groq AI client
from groq import Groq

# Import PDF reader library
from PyPDF2 import PdfReader

# -----------------------------------
# PAGE CONFIG
# -----------------------------------

# Configure Streamlit page settings
st.set_page_config(
    page_title="PDF QnA Assistant",   # Browser tab title
    page_icon="📄",                  # Browser tab icon
    layout="centered"               # Page layout
)

# -----------------------------------
# TITLE
# -----------------------------------

# Main heading
st.title("📄 PDF QnA Assistant")

# Small description below title
st.write("Upload PDF and ask questions using AI.")


# -----------------------------------
# SIDEBAR
# -----------------------------------

# Sidebar title
st.sidebar.title("⚙️ Settings")

# API key input
api_key = st.sidebar.text_input(
    "Enter Groq API Key",
    type="password"
)


# -----------------------------------
# PDF UPLOAD
# -----------------------------------

# Upload PDF file
uploaded_file = st.file_uploader(
    "Upload PDF",
    type="pdf"
)

# -----------------------------------
# PDF TEXT EXTRACTION
# -----------------------------------

# Function to extract text from uploaded PDF
def extract_text(pdf_file):

    # Empty string to store extracted text
    text = ""

    # Read PDF file
    reader = PdfReader(pdf_file)

    # Loop through all pages in PDF
    for page in reader.pages:

        # Extract text from current page
        page_text = page.extract_text()

        # Check if text exists
        if page_text:

            # Add page text to main text
            text += page_text + "\n"

    # Return full PDF text
    return text

# -----------------------------------
# SESSION STATE
# -----------------------------------

# Create chat history variable if not already present
if "chat_history" not in st.session_state:

    st.session_state.chat_history = []

# -----------------------------------
# MAIN APP
# -----------------------------------

# Check if user uploaded PDF
if uploaded_file is not None:

    # Show loading animation while reading PDF
    with st.spinner("Reading PDF..."):

        # Extract text from PDF
        document_text = extract_text(uploaded_file)

    # Success message after upload
    st.success("✅ PDF Uploaded Successfully")

    # Show total character count
    st.info(f"📑 Total Characters: {len(document_text)}")

    # Input box for user question
    question = st.text_input(
        "Ask a question from the PDF"
    )

    # Create Ask AI button
    ask_button = st.button("Ask AI")

    # Run when button clicked
    if ask_button:

        # Check API key entered or not
        if not api_key:

            st.error("❌ Please enter Groq API Key")

        # Check question entered or not
        elif not question:

            st.warning("⚠️ Please enter a question")

        else:

            try:

                # Create Groq client using API key
                client = Groq(api_key=api_key)

                # Create AI prompt
                prompt = f'''
You are a helpful PDF assistant.

PDF Content:
{document_text}

Question:
{question}

Answer clearly using ONLY the PDF content.
'''

                # Show thinking animation
                with st.spinner("Thinking..."):

                    # Send request to Groq AI
                    response = client.chat.completions.create(

                        # AI model name
                        model="llama-3.1-8b-instant",

                        # Chat messages
                        messages=[
                            {
                                "role": "user",
                                "content": prompt
                            }
                        ],

                        # AI creativity level
                        temperature=0.2
                    )

                    # Extract AI answer from response
                    answer = response.choices[0].message.content

                    # Save question and answer in history
                    st.session_state.chat_history.append({
                        "question": question,
                        "answer": answer
                    })

                    # Show answer heading
                    st.subheader("🤖 AI Answer")

                    # Display AI answer
                    st.write(answer)

            # Handle errors
            except Exception as e:

                # Show error message
                st.error(f"Error: {e}")

# -----------------------------------
# CHAT HISTORY
# -----------------------------------
```

```python
# Import threading library
# Used to run Streamlit in background
import threading

# Import os library
# Used to run terminal commands
import os

# Function to start Streamlit app
def run_streamlit():

    # Run Streamlit app from app.py file
    os.system("streamlit run app.py")

# Create a thread for Streamlit server
thread = threading.Thread(target=run_streamlit)

# Start the thread
# This launches Streamlit app
thread.start()
```
